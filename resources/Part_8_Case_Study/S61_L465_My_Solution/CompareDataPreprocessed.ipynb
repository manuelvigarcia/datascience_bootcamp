## Here is the bug
It is clear now that the lecture data is the wrong one: they interpret some of the 'Date' data as 'YYYY-dd-MM' (those that differ) while others as 'YYYY-MM-dd', as I did. Here is a sample of the rows that differ:

```
        Date 	Month 	DayOfWeek
```
---

```
5 	2015-07-10 	10 	2
8 	2015-07-06 	6 	6
19 	2015-08-05 	5 	4
20 	2015-08-12 	12 	1
21 	2015-08-03 	3 	6
``` 

It is evident that at index 5 with date 2015-07-10, the month shall be 7 and not 10.

### Day of the week
The same way, in the column DayOfWeek the lecture data interprets day as the second field in the date, only wen the third is 12 or lower to be a month. The trick in this case is that some of this missinterpreted dates happen to fall in the same day of the week as the correct one. But as an example, in the display of `their_mega_day[['Date','Month','DayOfWeek']]` below:

```
        Date 	Month 	DayOfWeek
```
---

```
5 	2015-07-10 	10 	2
8 	2015-07-06 	6 	6
19 	2015-08-05 	5 	4
20 	2015-08-12 	12 	1
21 	2015-08-03 	3 	6
``` 

The 2015-07-10 is interpreted as October 7th being a Wednesday ('2' with Monday being '0') instead of July 10th which is a Friday ('4')
