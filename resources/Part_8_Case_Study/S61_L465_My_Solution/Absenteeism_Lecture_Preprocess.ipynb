```python
import pandas as pd
```

```python
#Load data from .CSV: drop headers and 1st column (the ID).
raw_csv_data = pd.read_csv('../../../../statistics/python/Absenteeism/Absenteeism_data.csv', sep=',', header=0, index_col='ID',
                          dtype = {'ID': int,'Reason for Absence':int,'Transportation Expense':int,
                                    'Distance to Work':int,'Age':int,'Daily Work Load Average':float,'Body Mass Index':int,
                                    'Education':int,'Children':int,'Pets':int,'Absenteeism Time in Hours':int},
                           parse_dates=['Date'], date_format="%d/%m/%Y", decimal='.'
)
raw_csv_data.info
```

```python
raw_csv_data.describe()
```

```python
absence_groups = pd.DataFrame(raw_csv_data['Reason for Absence'].isin(range(1,15)).map({True:1,False:0}))
absence_groups['Reason_2'] = raw_csv_data['Reason for Absence'].isin(range(15,18)).map({True:1,False:0})
absence_groups['Reason_3'] = raw_csv_data['Reason for Absence'].isin(range(18,22)).map({True:1,False:0})
absence_groups['Reason_4'] = raw_csv_data['Reason for Absence'].isin(range(22,29)).map({True:1,False:0})
absence_groups['NoReason'] = (raw_csv_data['Reason for Absence'] == 0).map({True:1,False:0})
absence_groups
```

### Rename
- Get the column names
- Store the array in a variable

- Set objects columns to the renamed ones

```python
absence_groups.columns.values
```

```python
absence_columns = ['Reason_1', 'Reason_2', 'Reason_3', 'Reason_4',
       'NoReason']
absence_groups.columns = absence_columns
absence_groups.columns.values
```

### Reorder
- Get the column names
- Store the array in a variable
- Re-order the column names without changing the names
- Make the object equal to the object "of" the reordered column names

```python
absence_columns = ['NoReason', 'Reason_1', 'Reason_2', 'Reason_3', 'Reason_4']
absence_groups = absence_groups[absence_columns]
absence_groups
```
