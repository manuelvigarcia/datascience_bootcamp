```python
import pandas as pd
```

```python
#Load data from .CSV: drop headers and 1st column (the ID).
raw_csv_data = pd.read_csv('../../../../statistics/python/Absenteeism/Absenteeism_data.csv', sep=',', header=0,
                          dtype = {'ID': int,'Reason for Absence':int,'Transportation Expense':int,
                                    'Distance to Work':int,'Age':int,'Daily Work Load Average':float,'Body Mass Index':int,
                                    'Education':int,'Children':int,'Pets':int,'Absenteeism Time in Hours':int},
                           parse_dates=['Date'], date_format="%d/%m/%Y", decimal='.'
)
raw_csv_data.info
```

```python
raw_csv_data = raw_csv_data.drop(['ID'], axis=1)
```

```python
raw_csv_data.describe()
```

```python
raw_csv_data['Reason_1'] = raw_csv_data['Reason for Absence'].isin(range(1,15)).map({True:1,False:0})
raw_csv_data['Reason_2'] = raw_csv_data['Reason for Absence'].isin(range(15,18)).map({True:1,False:0})
raw_csv_data['Reason_3'] = raw_csv_data['Reason for Absence'].isin(range(18,22)).map({True:1,False:0})
raw_csv_data['Reason_4'] = raw_csv_data['Reason for Absence'].isin(range(22,29)).map({True:1,False:0})
raw_csv_data['None'] = (raw_csv_data['Reason for Absence'] == 0).map({True:1,False:0})
```

```python
raw_csv_data['Month']=raw_csv_data['Date'].dt.month
raw_csv_data['DayOfWeek']=raw_csv_data['Date'].dt.dayofweek
```

```python
raw_csv_data['EducationLevel'] = raw_csv_data['Education'].isin(range(2,5)).map({True:1,False:0})
```

```python
absenteeism_preprocessed = raw_csv_data[['Reason_1','Reason_2','Reason_3','Reason_4','None','Month','DayOfWeek','Transportation Expense',
                                         'Distance to Work','Age','Daily Work Load Average','Body Mass Index',
                                         'EducationLevel','Children','Pets','Absenteeism Time in Hours']].rename(columns={'EducationLevel':'Education'})
absenteeism_preprocessed.describe()
```

```python
absenteeism_preprocessed = absenteeism_preprocessed.drop(['None'], axis=1)
```
