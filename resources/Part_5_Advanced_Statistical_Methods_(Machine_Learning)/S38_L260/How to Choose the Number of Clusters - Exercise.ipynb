## Import the relevant libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import seaborn
seaborn.set_theme()
```

## Load the data

Load data from the csv file: <i> 'Countries_exercise.csv'</i>.

```python
raw_data = pd.read_csv('Countries-exercise.csv')
raw_data
```

Remove the duplicate index column from the dataset. 

Plot the <i>'Longtitude'</i> and <i>'Latitude'</i> columns. 

## Plot the data

Plot the <i>'Longtitude'</i> and <i>'Latitude'</i> columns. 

```python
plt.scatter(raw_data['Longitude'],raw_data['Latitude'])
plt.show()
```

## Select the features

Make sure to select the appropriate features since we are no longer using the categorical variable for our clustering but rather <i>'Longitude'</i> and <i>'Laditude'</i>.

```python
x = raw_data.loc[:,['Longitude','Latitude']]
x
```

## Clustering

Use 4 clusters initially.

```python
kmeans=KMeans(4)
kmeans.fit(x)
```

### Clustering results

```python
clusters = kmeans.fit_predict(x)
clusters
```

```python
data_with_clusters = raw_data.copy()
data_with_clusters['Cluster'] = clusters
```

Plot the data once again but separate the data by the clusters we defined.  

## Plot the data

```python
plt.scatter(data_with_clusters['Longitude'],data_with_clusters['Latitude'],c=data_with_clusters['Cluster'],cmap='rainbow')
plt.show()
```

## Selecting the number of clusters

### WCSS

Use the ingerated <i>sklearn</i> method <i> 'inertia_' </i>.

```python
kmeans.inertia_
```

Write a loop that calculates and saves the WCSS for any number of clusters from 1 up to 10 (or more if you wish).

```python
wcss=[]
for i in range(1,7):
    kmeans = KMeans(i)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)
```

```python
number_of_clusters=range(1,7)
plt.plot(number_of_clusters,wcss)
plt.show
```

### The Elbow Method

```python
for i in range(2,5):
    kmeans = KMeans(i)
    kmeans.fit(x)
    clusters = kmeans.fit_predict(x)
    data_w_clusters = raw_data.copy()
    data_w_clusters['Cluster']=clusters
    plt.scatter(data_w_clusters['Longitude'],data_w_clusters['Latitude'],c=data_w_clusters['Cluster'],cmap='rainbow')
    plt.show()
```
