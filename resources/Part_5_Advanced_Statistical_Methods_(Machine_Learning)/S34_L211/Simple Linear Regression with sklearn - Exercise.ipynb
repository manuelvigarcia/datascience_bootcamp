## Import the relevant libraries

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression as LR
sns.set()
```

## Load the data

```python
raw_data = pd.read_csv('real_estate_price_size.csv')
raw_data.describe()
```

## Create the regression

### Declare the dependent and the independent variables

```python
y = raw_data['price']
x1 = raw_data['size'].values.reshape(-1,1)
```

### Regression itself

```python
reg = LR()
reg.fit(x1,y)
```

### Calculate the R-squared

```python
reg.score(x1,y)
```

### Find the intercept

```python
reg.intercept_
```

### Find the coefficients

```python
reg.coef_
```

```python
yhat = 101912.60180122897 + 223.17874259*x1
plt.scatter(x1,y)
fig = plt.plot(x1,yhat, c='#880000')
plt.xlabel('size')
plt.ylabel('price')
```

### Making predictions

You find an apartment online with a size of 750 sq.ft.

All else equal what should be its price according to the model?

```python
new_sizes = [[750,1500,1250,650,900,1750,1105,1380,495]]
new_prices = reg.predict(np.array(new_sizes).reshape(-1,1))
new_prices
```
